```python
list1=[]
print("========== LIST 1 ==========")
for i in range(5):
    x=int(input("Enter a number:"))
    list1.append(x)
print(list1)
list2=[]
print("========== LIST 2 ==========")
for i in range(5):
    y=int(input("Enter a number:"))
    list2.append(y)
print(list2)
list3=[]
for i in range(5):
    list3.append(list1[i]+list2[i])
print()
print("===== LIST 1 + LIST 2 ======")
print(list3)
```
