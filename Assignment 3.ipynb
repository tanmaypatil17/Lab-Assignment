```python
a=int(input("Enter First Side:"))
b=int(input("Enter Second Side:"))
c=int(input("Enter Third Side:"))

def right_triangle():
    if a*a==b*b+c*c:
        print("Triangle is right-angled")
    elif b*b==a*a+c*c:
        print("Triangle is right-angled")
    elif c*c==a*a+b*b:
        print("Triangle is right-angled")
    else:
        print("Triangle is not right-angled")

right_triangle()
```
