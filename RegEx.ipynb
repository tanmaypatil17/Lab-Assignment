```python
import re
```

```python
txt = "The rain in Spain"
x = re.search("^The.Spain$",txt)
print(x)
if x:
    print("Match Found!")
else:
    print("No Match")
# " ^ " means starts with
# " . " means any character (except new line)
# " * " means zero or more occurence
# " $ " means ends with
```
