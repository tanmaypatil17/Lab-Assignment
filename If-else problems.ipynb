```python
age=int(input("Enter the age:"))
if age<=13:
    print("A person with that age would be defined as a Child")
elif age>=13 and age<=19:
    print("A person with that age would be defined as a Teenager")
elif age>=20 and age<=59:
    print("A person with that age would be defined as an Adult")
else:
    print ("A person with that age would be defined as a Senior Citizen")
```
