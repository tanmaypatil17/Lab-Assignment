```python
import pandas as pd
data = {"Name" : ["Tanmay","Rahul","Ayush","Aarav","Rahul"],
        "Marks" : [85,72,None,90,72],
        "City" : ["Pune","Mumbai","Pune","Delhi","Mumbai"]}

df = pd.DataFrame(data)
print("Original Data:")
print(df)
print("\nFirst 5 rows:")
print(df.head())
print("\nDataset Information")
df.info()
print(df.describe())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df = df.drop_duplicates()
result = df.loc[df["Marks"]>75]
print("\nAfter Data Preprocessing:")
print(df)
print("\nStudents with marks greater than 75:")
print(result)
```
