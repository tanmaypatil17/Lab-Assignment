```python
#List
students = ["Rohan","Isha","Kabir"]
students.append("Ananya")
students.insert(1,"Vihaan")
students.remove("Kabir")
students.pop(0)
students.sort()
print("List:",students)
```

```python
#Tuple
marks = (85,90,85,75)

print("Count:",marks.count(85))
print("Index:",marks.index(90))
print("Length:",len(marks))
print("Maximum:",max(marks))
print("Minimum:",min(marks))
```

```python
#Dictionary
student = {"Name" : "Tanmay", "Age" : 18, "Course" : "CSE AIDS"}

print(student.get("Name"))
print(student.keys())
print(student.values())

student.update({"Age" : 19})
student.pop("Course")

print("Dictionary:",student)
```
