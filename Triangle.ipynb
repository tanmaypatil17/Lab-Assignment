```python
i=1
while i==1:
    s1=int(input("Enter first side:"))
    s2=int(input("Enter Second side:"))
    s3=int(input("Enter Third Side:"))
    def notvalid():
        return s1+s2<s3 or s2+s3<s1 or s3+s1<s2
    def equi():
        return s1==s2==s3
    def iso():
        return s1==s2 or s2==s3 or s3==s1
    def right():
        return s1**2 + s2**2 == s3**2 or s1**2 + s3**2 == s2**2 or s2**2 + s3**2 == s1**2
    def scal():
        return s1!=s2!=s3
    if notvalid()==True:
        print("Sides are invalid to create a triangle")
        break
    elif equi()==True:
        print("The triangle is equilateral")
    elif iso()==True:
        print("The triangle is isosceles")
    elif right()==True and scal()==True:
        print("The triangle is right angled \nThe triangle is also scalene")
    elif right()==True:
        print("The triangle is right angled")
    else:
        print("The Triangle is Scalene")
    cont=input("Do you want to continue? (Yes/No):")
    if cont=="Yes" or cont=="yes":
        print("Ahh shit! Here we go again")
        continue
    elif cont=="No" or cont=="no":
        print("Shutting off.....")
        break
    else:
        print("Invalid! Please do as directed")
```
